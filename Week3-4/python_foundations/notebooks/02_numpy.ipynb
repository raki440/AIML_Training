## 1. Arrays: shape, dtype, math on the whole array

```python
a = np.array([10, 20, 30, 40])
print(a.shape, a.dtype, a.mean(), a * 2, a + a)
```

## 2. Vectorization: no Python loops
Same result, very different speed.

```python
x = np.arange(1_000_000)
t = time.perf_counter(); s1 = sum(i * 2 for i in range(1_000_000)); loop = time.perf_counter() - t
t = time.perf_counter(); s2 = (x * 2).sum(); vec = time.perf_counter() - t
print(s1 == s2, f"python loop {loop:.3f}s | numpy {vec:.4f}s")
```

## 3. 2D arrays and axis
`axis=0` goes down the rows (per column), `axis=1` goes across (per row).

```python
m = np.array([[1, 2, 3], [4, 5, 6]])
print(m.shape, m.T.shape)
print("per column:", m.sum(axis=0), "| per row:", m.sum(axis=1))
```

## 4. Broadcasting: standardize every column in one line
**Reusable idea:** scaling features. scikit-learn's `StandardScaler` does this for you later.

```python
rng = set_seed(42)
data = rng.normal([50, 70], [10, 20], size=(5, 2))   # 5 rows: tenure-like, charges-like
scaled = (data - data.mean(axis=0)) / data.std(axis=0)
print(scaled.round(2)); print(scaled.mean(axis=0).round(6), scaled.std(axis=0).round(6))
```

## 5. Boolean masks and missing values

```python
charges = np.array([55.0, 72.5, np.nan, 130.0, 61.2])
print("NaN count:", np.isnan(charges).sum(), "| mean ignoring NaN:", np.nanmean(charges))
print("above 100:", charges[charges > 100])
print("clipped:", np.clip(charges, 0, 100))
```

## 6. Random numbers with a seed
**Reusable:** `set_seed()` makes experiments repeatable, which is the first rule of reproducible ML.

```python
print(set_seed(42).integers(0, 100, 5), set_seed(42).integers(0, 100, 5))   # identical
print(np.random.default_rng().integers(0, 100, 5))                          # different each run
```
