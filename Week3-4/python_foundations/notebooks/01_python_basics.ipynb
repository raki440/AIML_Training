## 1. Types and collections
list = ordered, dict = key to value. Config, rows and API payloads are all dicts and lists.

```python
tenure = 12; rate = 0.35; plan = "basic"; active = True
plans = ["basic", "standard", "premium"]
customer = {"id": 1, "plan": "basic", "tenure_months": 12}
print(type(tenure), type(rate), type(plan), type(active))
print(plans[0], customer["plan"], len(plans))
```

## 2. Control flow and comprehensions

```python
squares = [x**2 for x in range(5)]
without_id = {k: v for k, v in customer.items() if k != "id"}
for p in plans:
    print(p, "-> entry plan" if p == "basic" else "")
print(squares, without_id)
```

## 3. Functions with type hints and docstrings
**Reusable:** every `mlkit` function is written this way. Hints document inputs and outputs.

```python
def churn_risk(tenure_months: int, tickets: int) -> str:
    """Rule-of-thumb risk label (teaching example, not a model)."""
    if tenure_months < 6 and tickets >= 3:
        return "high"
    return "medium" if tickets >= 3 else "low"

print(churn_risk(3, 4), churn_risk(40, 4), churn_risk(40, 0))
```

## 4. Errors: fail early, with a clear message
**Reusable:** `read_table` and `validate` raise clear errors instead of failing later.

```python
def safe_ratio(a: float, b: float) -> float:
    try:
        return a / b
    except ZeroDivisionError:
        return float("nan")

print(safe_ratio(6, 3), safe_ratio(1, 0))

try:
    open("missing_file.csv")
except FileNotFoundError as e:
    print("caught:", e)
```

## 5. Files and paths with pathlib, config as JSON
**Reusable:** `mlkit/config.py` keeps all paths in one place.

```python
from pathlib import Path
import json

cfg_path = Path("demo_config.json")
cfg_path.write_text(json.dumps({"seed": 42, "test_size": 0.2}))
cfg = json.loads(cfg_path.read_text())
print(cfg, cfg_path.suffix, cfg_path.exists())
cfg_path.unlink()
```
